```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
text = "The students are studying, playing, and running while the cats were eating."

words = word_tokenize(text)
print("Original Text:", text)
print("Tokenized Words:", words)
stemmer = PorterStemmer()

print("Stemming Results:")
for word in words:
    if word.isalpha():
        print(word, "->", stemmer.stem(word.lower()))
lemmatizer = WordNetLemmatizer()

print("Lemmatization Results:")
for word in words:
    if word.isalpha():
        print(word, "->", lemmatizer.lemmatize(word.lower()))
print("Lemmatization with POS:")

for word in words:
    if word.isalpha():
        print(word, "->", lemmatizer.lemmatize(word.lower(), pos='v'))
```
